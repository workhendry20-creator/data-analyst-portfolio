
# Sales Data Analysis Portfolio

This project analyzes sales performance using Python and Google Colab. The analysis includes data cleaning, exploratory data analysis (EDA), and visualization to identify sales trends and top-performing product categories.

**Import Libraries**

```python
import pandas as pd
import matplotlib.pyplot as plt
```

**Load Dataset**

The dataset contains sales transaction data including product categories, sales amount, customer information, and order details.

```python
df = pd.read_csv('/content/sales_data_sample.csv', encoding='latin1')
df.head()
```

**Data Cleaning**

This step checks missing values and removes duplicate data to ensure accurate analysis.

```python
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
total_sales = df['SALES'].sum()
print("Total Sales:", total_sales)
```

**Exploratory Data Analysis (EDA)**

The analysis focuses on identifying top-performing product lines based on total sales.

```python
top_product = df.groupby('PRODUCTLINE')['SALES'].sum().sort_values(ascending=False)

print(top_product)
```

**Visualization**

Visualization helps identify product categories with the highest sales contribution.

```python
top_product.plot(kind='bar')

plt.title('Sales by Product Line')
plt.xlabel('Product Line')
plt.ylabel('Sales')

plt.show()
```

## Key Insights

- Classic Cars generated the highest total sales.
- Vintage Cars ranked as the second best-performing category.
- Trains contributed the lowest sales among all product lines.
- Product sales distribution indicates customer preference trends.
